## 🧩Paso 4: Summary statistics de uso por usuario


### 4.1 Agrupación por comportamiento de uso

🎯**Objetivo**: Resumir las variables clave de la tabla `usage` **por usuario**, creando métricas que representen su comportamiento real de uso histórico. 

**Instrucciones:**: 
1. Construye una tabla agregada de `usage` por `user_id` que incluya:
- número total de mensajes  
- número total de llamadas  
- total de minutos de llamadas

2. Renombra las columnas para que tengan nombres claros:  
- `cant_mensajes`  
- `cant_llamadas`  
- `cant_minutos_llamada`
3. Combina esta tabla con `users`.

```python

# Columnas auxiliares
usage["is_text"] = (usage["type"] == "text").astype(int) #conocer el total de mensajes
usage["is_call"] = (usage["type"] == "call").astype(int) #conocer el total de llamadas


# Agrupar información por usuario
usage_agg = usage.groupby("user_id").agg(
    cant_mensajes=("is_text", "sum"),
    cant_llamadas=("is_call", "sum"),
    cant_minutos_llamada=("duration", "sum")
).reset_index()

# Observar resultado
usage_agg.head(3)

```

<div class="alert alert-block alert-success">
<b>✅ Comentario del revisor. (Iteración 1)</b> <a class="tocSkip"></a><br>
Agregación por usuario correcta: las columnas auxiliares con <code>sum</code> cuentan mensajes y llamadas, y <code>duration</code> suma los minutos.
</div>

<div class="alert alert-block alert-info">
<b>💡 Comentario del revisor. (Iteración 1) — Tip de eficiencia</b> <a class="tocSkip"></a><br>
Con <i>named aggregation</i> agrupas y renombras en un solo paso: <code>.agg(cant_mensajes=('is_text','sum'), ...)</code>.
</div>

```python

# Renombrar columnas
usage_agg = usage_agg.rename(columns=
{ "is_text": "cant_mensajes",
  "is_call": "cant_llamadas",
  "duration": "cant_minutos_llamada"  })
# observar resultado
usage_agg.head(3)
```

```python
# Combinar las tablas conservando todos los usuarios
user_profile = pd.merge(
    users,
    usage_agg,
    on='user_id',
    how='left')
columnas_uso = [
    'cant_mensajes',
    'cant_llamadas',
    'cant_minutos_llamada']
user_profile[columnas_uso] = user_profile[columnas_uso].fillna(0)
# Verificar el resultado
user_profile.head(5)
```

<div class="alert alert-block alert-danger">
<b>❌ Comentario del revisor. (Iteración 1) — Necesita arreglos</b> <a class="tocSkip"></a><br>
Revisa el tipo de unión: sin el parámetro <code>how</code>, <code>pd.merge</code> hace un <b>inner join</b> y solo conserva a los clientes que tienen registros en <code>usage</code>. Por eso tu <code>describe()</code> muestra <code>count</code> = <b>3999</b> en lugar de 4000: se perdió un cliente, y eso cambia los porcentajes de plan y de segmentos.<br><br>

### 6.1 Segmentación de Clientes Por Uso

🎯 **Objetivo:** Clasificar a cada usuario en un grupo de uso (Bajo uso, Uso medio, Alto uso) basándose en la cantidad de llamadas y mensajes registrados.

**Instrucciones:**  
- Crea una nueva columna llamada `grupo_uso` en el dataframe `user_profile`.
- Usa comparaciones lógicas (<, >) para evaluar las condiciones de llamadas y mensajes y asigna:
  - `'Bajo uso'` cuando llamadas < 5 y mensajes < 5
  - `'Uso medio'` cuando llamadas < 10 y mensajes < 10
  - `'Alto uso'` para el resto de casos

```python

# Crear columna grupo_uso

variables = [
    (user_profile['cant_llamadas'] < 5) & (user_profile['cant_mensajes'] < 5),
    (user_profile['cant_llamadas'] < 10) & (user_profile['cant_mensajes'] < 10)]
grupo_uso = ['Bajo uso', 'Uso medio']
user_profile['grupo_uso'] = np.select(
    variables,
    grupo_uso,
    default='Alto uso'
)
```
